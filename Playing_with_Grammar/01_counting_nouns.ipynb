# **Counting nouns – plural and singular nouns**


In this recipe, we will do two things:

- Determine whether a noun is plural or singular

- Turn plural nouns into singular nouns and vice versa

You might need these two things in a variety of tasks: in making your chatbot speak in grammatically correct sentences,

in coming up with text classification features, and so on.


https://pypi.org/project/inflect/

````python
import nltk
import inflect
from nltk import pos_tag
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
````

Number detection works in one of two ways.

1. One is by reading the **part of speech tag assigned by NLTK**. 
    - If the tag is NN, then the noun is singular, and if it is NNS, then it's
plural. 

2. The other way is to **use the WordNet lemmatizer** and to compare the lemma and
the original word. 
    - The noun is singular if the lemma and the original input noun are the
same, and plural otherwise

To find the singular form of a plural noun and the plural form of a singular noun, we
can use the inflect package. 

Its ```plural``` and ```singular_noun``` methods return the
correct forms.

````python
# Read file
filename = '..\Chapter01\sherlock_holmes_1.txt'
file = open(filename, 'r', encoding='utf-8')
sherlock_homes_text = file.read()

sherlock_homes_text
````

````python
# Remove new lines
sherlock_homes_text = sherlock_homes_text.replace('\n', ' ')
sherlock_homes_text
````

````python
# Tokenize
token = word_tokenize(sherlock_homes_text)

# pos tagging
words_with_pos = pos_tag(token)
````

#### Define the get_nouns function, which will filter out the nouns from all the words:

````python
def get_nouns(words_with_pos):
    noun_set = ['NN', 'NNS']
    nouns = [word for word in words_with_pos if word[1] in noun_set]
    return nouns

nouns = get_nouns(words_with_pos)
print(nouns)
````

To determine whether a noun is singular or plural, we have two options.

1. Detect plural nouns with NLTK.

````python
def is_plural_nltk(nouns):
    pos = nouns[1]
    if pos == 'NNS':
        return True
    else:
        return False
    
plural_nouns = is_plural_nltk(nouns)
print(plural_nouns)
````

2. Detect plural nouns with Lemmatizer

````python
from nltk.stem import WordNetLemmatizer

def is_plural_wn(nouns):
    wnl = WordNetLemmatizer()
    results = []

    for word, tag in nouns:
        lemma = wnl.lemmatize(word, 'n')
        results.append((word, word != lemma))

    return results


plural_nouns = is_plural_wn(nouns)
print(plural_nouns)
````

### Coverting Singlular --> Plural

````python
def get_plural(singular_nouns):
    p = inflect.engine()
    return p.plural(singular_nouns)

get_plural('woman')
get_plural('bat')
````

### Coverting Plural --> Singular

````python
def get_singular(plural_nouns):
    p=inflect.engine()
    return p.singular_noun(plural_nouns)

get_singular('women')
get_singular('players')
````

### Flip plural ↔ singular for all nouns

We can now use the two preceding functions to return a list of nouns changed into plural or singular, depending on the original noun. The following code uses the ```is_plural_wn``` function to determine if the noun is plural. You can also use the ```is_plural_nltk``` function:

````python
# Exception, the above code works as well, but not here. So, use this atm.
def is_plural_wn(noun):
    wnl = WordNetLemmatizer()
    lemma = wnl.lemmatize(noun, 'n')
    plural = True if noun is not lemma else False
    return plural
````

စကားလုံးအသစ်တွေထုတ်သလိုသုံးလို့ရတယ်။

ဥပမာ plural ရှိတယ်ဆိုရင် singular ပြောင်းပြီး word list ထဲ အသစ်ထည့်၊ နဂို singular ရှိတယ်ဆိုလည်း plural ပြောင်းပြီး ထည့်။

````python
def plurals_wn(words_with_pos):
    other_nouns = []
    for noun_info in words_with_pos:
        word = noun_info[0]
        plural = is_plural_wn(word)
        if (plural):
            singular = get_singular(word)
            other_nouns.append(singular)
        else:
            plural = get_plural(word)
            other_nouns.append(plural)
    return other_nouns

other_nouns = plurals_wn(words_with_pos)
print(other_nouns)
````
