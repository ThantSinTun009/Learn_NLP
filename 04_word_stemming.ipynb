# Word stemming

In some NLP tasks, we need to stem words, or remove the suffixes and endings such as **-ing** and **-ed**. 

To do this recipe, we will be using NLTK and its **Snowball Stemmer**.

```python
from nltk.stem.snowball import SnowballStemmer
```

```python
# Initialize stemmer with English
stemmer = SnowballStemmer('english')
stemmer
```

```python
# initialize word list to stem
words = ['leaf', 'leaves', 'booking', 'writing', 'completed', 'stemming', 'skies']
```

```python
# stem words
stemmed_words = [stemmer.stem(word) for word in words]
stemmed_words
```

> The stemmer strips suffixes and endings from the words. As seen here, it removes suffixes such as -es, -ing, and -ed, as well as others, such as -ive, -ize, and -ment. It also handles exceptions, as seen in the case of skies → sky. However, it does not change the stem to its canonical form, as the example with the word leaves shows.

```python
# NLTK Snowball has algorithms for sevreal langauges
print(SnowballStemmer.languages)
```

```python
# Eg: Spanish
stemmer = SnowballStemmer('spanish')
spanish_words = ['caminando', 'amigo', 'bueno']
stemmed_words = [stemmer.stem(word) for word in spanish_words]

stemmed_words
```
