# **Removing stopwords**

When we work with words, especially if we are considering words' semantics, we sometimes need to exclude some **very frequent words that do not bring any substantial meaning to a sentence**, words such as but, can, we, and so on. This recipe shows how to do that.

For this recipe, we will need a list of stopwords to be defined or can also use the stopwords list provided with the nltk package.

> Stopwords are very common words that usually don’t add much meaning to text.

👉 We remove these words so the model can focus on important words.

```python
import nltk
import csv
from nltk.tokenize import word_tokenize
```

```python
# Initialize the stopwords list:
csv_file="stopwords.csv"

# read stopwords list
with open(csv_file, 'r', encoding='utf-8') as fp:
 reader = csv.reader(fp, delimiter=',', quotechar='"')
 stopwords = [row[0] for row in reader]
```

```python
# alternatively, stopword list from nltk
stopwords = nltk.corpus.stopwords.words('english')
```

**Information**
> Here is a list of languages that NLTK supports for stopwords: Arabic,
Azerbaijani, Danish, Dutch, English, Finnish, French, German, Greek,
Hungarian, Italian, Kazakh, Nepali, Norwegian, Portuguese, Romanian,
Russian, Spanish, Swedish, and Turkish.

```python
# 4. Read in the text file:
filename = 'sherlock_holmes_1.txt'
file = open(filename, 'r', encoding='utf-8')
text = file.read()
```

```python
# preprocess
text = text.replace("\n", " ")

# tokenize
words = word_tokenize(text)
```

**Remove the stopwords**

```python
# Remove the stopwords:
words = [word for word in words if word.lower() not in stopwords]
words
```

## We can also compile a stopwords list by using the text we are working with and calculating the frequencies of the words in it.

> **The FreqDist object** in the NLTK package counts the number of occurrences of each word:

```python
import nltk
from nltk.probability import FreqDist

file = open(filename, "r", encoding="utf-8")
text = file.read()

text = text.replace("\n", " ")

words = word_tokenize(text)
```

Create the frequency distribution object and use it to create a list of tuples where
- the first element is the word and
- the second one is the frequency count

```python
freq_dist = FreqDist(word.lower() for word in words)

words_with_frequencies = [(word, freq_dist[word]) for word in freq_dist.keys()]
```

```python
freq_dist
```

```python
words_with_frequencies
```

```python
# Sort the list of tuples by frequency
sorted_words = sorted(words_with_frequencies, key=lambda tup: tup[1])
sorted_words
```

Now we have two options:
> 1. use a frequency cutoff for the stopwords or


> 2. take the top n% of words sorted by frequency.

Here is the first option. Use 100 as the frequency cutoff

```python
stopwords = [tuple[0] for tuple in sorted_words if tuple[1] > 100 ]
stopwords
```

frequency အကြိမ် ၁၀၀ ကျော်တဲ့ word မပါတော့ stopword မရှိသလိုဖြစ်သွား။

The other option is to use the n% most frequent words as stopwords. Here I use 0.2% of the most frequent words:

```python
length_cutoff = int(0.02*len(sorted_words))
stopwords = [tuple[0] for tuple in sorted_words[-length_cutoff:]]

stopwords
```
