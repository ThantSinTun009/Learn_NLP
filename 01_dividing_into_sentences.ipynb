```python
import time
import nltk
import spacy
```

```python
# Import tokenizer
tokenizer = nltk.data.load("tokenizers/punkt/english.pickle")

# not just tokenizer, it is the whole nlp pipeline
nlp = spacy.load("en_core_web_sm")
```

- NLTK loads a tokenizer directly, spaCy loads a pipeline that includes a tokenizer.

```python
# Read text file
filename = "sherlock_holmes_1.txt"
file = open(filename, "r", encoding="utf-8")
text = file.read()

text
```

```python
# Replace newlines with spaces -- Preprocess
text = text.replace("\n", " ")
text
```

```python
# Divide the text into sentences with spaCy
doc = nlp(text)
sentences = [sentence.text for sentence in doc.sents]

sentences
```
